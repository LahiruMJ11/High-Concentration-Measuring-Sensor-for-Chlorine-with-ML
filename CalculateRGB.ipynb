```python
import cv2
import numpy as np

# Step 1: Read the image
image = cv2.imread("chlorine_sample.jpg")

# Convert from BGR (OpenCV default) to RGB
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

# Step 2: Define Region of Interest (ROI)
# Example: manually crop the center region (adjust based on setup)
h, w, _ = image_rgb.shape
roi = image_rgb[h//4: 3*h//4, w//4: 3*w//4]   # crop center 50%

# Step 3: Compute average R, G, B
mean_rgb = roi.mean(axis=(0,1))   # average over height & width
R_mean, G_mean, B_mean = mean_rgb

print("Average RGB:", R_mean, G_mean, B_mean)

# Step 4: Normalize if required
R_norm, G_norm, B_norm = R_mean/255, G_mean/255, B_mean/255
print("Normalized RGB:", R_norm, G_norm, B_norm)

# These values can now go into your model:
# model.predict([[R_norm, G_norm, B_norm]])
```
